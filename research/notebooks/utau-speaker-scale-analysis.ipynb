# UTAU 話者スケール分析

`research/scripts/utau_estimate_speaker_scale.py` が生成した話者別 `tractScale` 候補を確認するためのノートブック。

```python
import csv
from pathlib import Path

import matplotlib.pyplot as plt

cwd = Path.cwd()
if cwd.name == "notebooks":
    ROOT = cwd.parent
elif (cwd / "data" / "processed").exists():
    ROOT = cwd
else:
    ROOT = cwd / "research"

scale_path = ROOT / "data" / "processed" / "analysis" / "utau-speaker-scale.csv"
with scale_path.open(encoding="utf-8", newline="") as f:
    rows = list(csv.DictReader(f))

rows
```

```python
labels = [row["character_name"] or row["voice_id"] for row in rows]
tract_scales = [float(row["tract_scale"]) for row in rows]

plt.figure(figsize=(10, 5))
plt.bar(labels, tract_scales)
plt.axhline(1.0, color="black", linewidth=1)
plt.xticks(rotation=45, ha="right")
plt.ylabel("tractScale")
plt.title("UTAU speaker tractScale candidates")
plt.tight_layout();
```
