# UTAU ノイズ・子音成分分析

`research/scripts/utau_analyze_noise_components.py` が生成した摩擦音、息、子音、おまけ音声のスペクトル指標を確認するためのノートブック。

```python
import csv
import json
from pathlib import Path

import matplotlib.pyplot as plt

cwd = Path.cwd()
if cwd.name == "notebooks":
    ROOT = cwd.parent
elif (cwd / "data" / "processed").exists():
    ROOT = cwd
else:
    ROOT = cwd / "research"

analysis_dir = ROOT / "data" / "processed" / "analysis"
exports_dir = ROOT / "data" / "processed" / "exports"
details_path = analysis_dir / "utau-noise-components.csv"
summary_path = analysis_dir / "utau-noise-components-summary.csv"
label_summary_path = analysis_dir / "utau-noise-components-by-label.csv"
preset_path = exports_dir / "noise-component-presets.generated.json"

with details_path.open(encoding="utf-8", newline="") as f:
    details = list(csv.DictReader(f))
with summary_path.open(encoding="utf-8", newline="") as f:
    summary = list(csv.DictReader(f))
with label_summary_path.open(encoding="utf-8", newline="") as f:
    by_label = list(csv.DictReader(f))
presets = json.loads(preset_path.read_text(encoding="utf-8"))

len(details), summary, presets["presets"]
```

```python
target_labels = {"さ", "し", "す", "せ", "そ", "は", "ひ", "ふ", "へ", "ほ"}
rows = [row for row in by_label if row["label"] in target_labels]
rows = sorted(rows, key=lambda row: float(row["spectral_centroid_hz_median"]))
labels = [row["label"] for row in rows]
centroids = [float(row["spectral_centroid_hz_median"]) for row in rows]

plt.figure(figsize=(10, 4))
plt.bar(labels, centroids)
plt.ylabel("Hz")
plt.title("Noise component spectral centroid by label")
plt.tight_layout();
```

```python
rows = [row for row in by_label if row["label"] in target_labels]
rows = sorted(rows, key=lambda row: float(row["high_band_ratio_median"]))
labels = [row["label"] for row in rows]
high_ratios = [float(row["high_band_ratio_median"]) for row in rows]

plt.figure(figsize=(10, 4))
plt.bar(labels, high_ratios)
plt.ylabel("ratio")
plt.title("High-band ratio by label")
plt.tight_layout();
```
