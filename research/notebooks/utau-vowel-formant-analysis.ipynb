# UTAU 母音フォルマント分析

`research/scripts/utau_analyze_vowels.py` が生成した CSV と散布図を確認するためのノートブック。音声データ本体は Git 管理外の `research/data/raw/reference/utau-samples/` に配置する。

```python
import csv
from collections import Counter
from pathlib import Path

import matplotlib.pyplot as plt

cwd = Path.cwd()
if cwd.name == "notebooks":
    ROOT = cwd.parent
elif (cwd / "data" / "processed").exists():
    ROOT = cwd
else:
    ROOT = cwd / "research"
analysis_dir = ROOT / "data" / "processed" / "analysis"
details_path = analysis_dir / "utau-vowel-formants.csv"
summary_path = analysis_dir / "utau-vowel-formants-summary.csv"
plot_path = analysis_dir / "utau-vowel-formant-scatter.png"

with details_path.open(encoding="utf-8", newline="") as f:
    details = list(csv.DictReader(f))
with summary_path.open(encoding="utf-8", newline="") as f:
    summary = list(csv.DictReader(f))

details[:5]
```
