# UTAU 声質差分析

`research/scripts/utau_analyze_voice_quality.py` が生成した声質指標を確認するためのノートブック。スペクトル重心、高域比、スペクトル傾斜、非周期成分の近似比などを音源サブセット単位で比較する。

```python
import csv
from pathlib import Path

import matplotlib.pyplot as plt

cwd = Path.cwd()
if cwd.name == "notebooks":
    ROOT = cwd.parent
elif (cwd / "data" / "processed").exists():
    ROOT = cwd
else:
    ROOT = cwd / "research"

analysis_dir = ROOT / "data" / "processed" / "analysis"
details_path = analysis_dir / "utau-voice-quality.csv"
summary_path = analysis_dir / "utau-voice-quality-summary.csv"

with details_path.open(encoding="utf-8", newline="") as f:
    details = list(csv.DictReader(f))
with summary_path.open(encoding="utf-8", newline="") as f:
    summary = list(csv.DictReader(f))

summary
```

```python
def plot_metric(metric, title, ylabel):
    rows = sorted(summary, key=lambda row: float(row[metric]))
    labels = [row["voice_id"] for row in rows]
    values = [float(row[metric]) for row in rows]
    plt.figure(figsize=(12, 5))
    plt.bar(labels, values)
    plt.xticks(rotation=55, ha="right")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.tight_layout()

plot_metric("spectral_centroid_hz_median", "Spectral centroid by subset", "Hz")
```

```python
plot_metric("high_band_ratio_median", "High-band energy ratio by subset", "ratio")
```

```python
plot_metric("nonharmonic_energy_ratio_median", "Non-harmonic energy ratio by subset", "ratio")
```
